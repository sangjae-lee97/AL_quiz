## Q001

### 현재 내가 이해한 내용

2차원 리스트는 행과 열로 접근할 수 있다는 것은 이해했다.

### AI에게 한 질문

나는 흑백 사진을 압축하려고 해. 근데 답을 알려주지는 말고 내가 질문할테니 단계별로 차근차근 알려줘. 2차원 이미지에서 행과 열은 어떻게 표현되는가?

### AI 답변에서 이해한 핵심

중첩 반복문을 사용해 행과 열을 차례대로 확인할 수 있다.

### 내가 직접 해본 것

2×2 영역을 종이에 적고 모든 좌표를 직접 확인했다.

### 결과

영역 검사에 필요한 좌표 범위를 이해했다.
