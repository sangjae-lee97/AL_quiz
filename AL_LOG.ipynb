## Q008

### 현재 내가 이해한 내용

좌상단 부분은 시작 좌표가 변하지 않음을 확인

### AI에게 한 질문

오른쪽 위 영역의 시작 좌표는 어떻게 될까?

### AI 답변에서 이해한 핵심

* 원래 N×N 영역의 시작 좌표가 (r, c)이고, 절반 크기를 half = N / 2라고 할 때:오른쪽 위(우상단) 영역의 시작 좌표는 (r, c + half)가 됩니다.
* 이유:행(세로 위치): 왼쪽 위 사각형과 높이(위쪽 줄)가 같으므로 행 좌표는 그대로 r입니다.
* 열(가로 위치): 왼쪽 영역의 너비(half)만큼 오른쪽으로 건너뛰어야 하므로 열 좌표는 c + half가 됩니다.

### 내가 직접 해본 것

4×4 영역에서 실제로 4등분 한 뒤에, 원래 시작 영역의 좌표와 4등분 뒤의 우상단 부분 시작 좌표를 확인해봄

### 결과

원래 영역 시작 좌표 (r, c) != 우상단 영역 시작 좌표 (r, c + half) 임을 확인
